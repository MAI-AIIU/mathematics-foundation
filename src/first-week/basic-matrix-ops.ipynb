## 1: Write a function my_make_lin_ind(A), where A and B are matrices. Let the rank(A)=n. Then B should be a matrix containing the first n columns of A that are all linearly independent. Note that this implies that B is full rank.

```python
import numpy as np

def my_make_lin_ind(A):
  # create zero matrix, later we fill that with only independent columns
  B = np.empty((A.shape[0], 0))

  for i in range(A.shape[1]):
      column = A[:, i:i+1]

      new_B = np.hstack((B, column))

      old_rank = 0 if B.shape[1] == 0 else np.linalg.matrix_rank(B)
      new_rank = np.linalg.matrix_rank(new_B)

      if new_rank > old_rank:
          B = new_B

  return B
```

```python
A = np.array([
    [1, 2, 3],
    [5, 3, 1]
])

B = my_make_lin_ind(A)

print(B)
```
