# Inner Products, Orthogonal Projection, and Rotation

This lab illustrates three key ideas from linear algebra and machine learning:

1. Inner product of functions: we measure similarity between functions using an integral, not just a dot product.
2. Orthogonal projection: we approximate a function by the closest element in a chosen subspace.
3. Rotation: a rotation preserves lengths and angles, and therefore preserves inner products.

These ideas are foundational for approximation, Fourier analysis, and geometric reasoning in machine learning.

## 1) Inner Product of Functions

For two functions $f(x)$ and $g(x)$ defined on an interval $[a, b]$, the inner product is

$
\langle f, g \rangle = \int_a^b f(x)g(x)\,dx.
$

This generalizes the usual dot product in Euclidean space. Functions are orthogonal if their inner product is zero.

Example: over $[-\pi, \pi]$, the functions $\sin(x)$ and $\cos(x)$ are orthogonal because

$
\int_{-\\pi}^{\\pi} \sin(x)\cos(x)\,dx = 0.
$$

```python
import numpy as np

# Define an interval
x = np.linspace(-np.pi, np.pi, 2000)

# Example functions
f = np.sin(2 * x)
g = np.cos(3 * x)

# Inner product via numerical integration (trapezoidal rule)
inner_fg = np.trapezoid(f * g, x)
print("Inner product <sin(2x), cos(3x)> =", inner_fg)

# Check orthogonality between sin(x) and cos(x)
s = np.sin(x)
c = np.cos(x)
inner_sc = np.trapezoid(s * c, x)
print("Inner product <sin(x), cos(x)> =", inner_sc)

# Check orthogonality between sin(x) and 1
one = np.ones_like(x)
inner_s1 = np.trapezoid(s * one, x)
print("Inner product <sin(x), 1> =", inner_s1)
```

## 2) Orthogonal Projection

Suppose we want to approximate a function $f(x)$ using a subspace spanned by basis functions, for example ${1, \cos(x), \sin(x)}$.

The orthogonal projection of $f$ onto this subspace is the function closest to $f$ in the sense of minimizing the squared error

$
\|f - p\|^2 = \int_a^b (f(x)-p(x))^2 dx.
$

If the basis functions are orthogonal, the projection coefficients are easy to compute:

$
c_i = \frac{\langle f, e_i \rangle}{\langle e_i, e_i \rangle}.
$

This is the same idea as projecting vectors onto coordinate axes in Euclidean space.

```python
import numpy as np
import matplotlib.pyplot as plt

# Interval and target function
x = np.linspace(-np.pi, np.pi, 2000)
f = x  # Example function on [-pi, pi]

# Basis functions
e0 = np.ones_like(x)
e1 = np.cos(x)
e2 = np.sin(x)

# Compute projection coefficients
c0 = np.trapezoid(f * e0, x) / np.trapezoid(e0 * e0, x)
c1 = np.trapezoid(f * e1, x) / np.trapezoid(e1 * e1, x)
c2 = np.trapezoid(f * e2, x) / np.trapezoid(e2 * e2, x)

proj = c0 * e0 + c1 * e1 + c2 * e2

print("Projection coefficients:")
print("c0 =", c0)
print("c1 =", c1)
print("c2 =", c2)

# Visualization
plt.figure(figsize=(10, 6))
plt.plot(x, f, label="f(x) = x", linewidth=2)
plt.plot(x, proj, label="Orthogonal projection onto span{1, cos x, sin x}",
         linestyle="--", linewidth=2)
plt.axhline(0, color="black", linewidth=0.8)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Orthogonal Projection of f(x)=x onto a Trigonometric Basis")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Error norm
error = np.trapezoid((f - proj) ** 2, x)
print("Squared projection error =", error)
```

## 3) Rotation

A rotation in 2D is represented by a matrix

$
R(\theta) =
\begin{bmatrix}
\cos\theta & -\sin\theta\\
\sin\theta & \cos\theta
\end{bmatrix}.
$
<br>Rotation preserves the length of vectors and the angle between vectors. In particular, it preserves the dot product:

$
R u \cdot R v = u \cdot v.
$

This is why rotations are important in geometry and in many optimization and machine learning pipelines.

```python
import numpy as np
import matplotlib.pyplot as plt

# Rotation angle
theta = np.pi / 6
R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

# Original vector
v = np.array([1.0, 0.0])
rotated_v = R @ v
print("Original vector v =", v)
print("Rotated vector Rv =", rotated_v)

# Check length preservation
print("||v|| =", np.linalg.norm(v))
print("||Rv|| =", np.linalg.norm(rotated_v))

# Rotate a unit circle
t = np.linspace(0, 2 * np.pi, 500)
circle = np.column_stack([np.cos(t), np.sin(t)])
rotated_circle = circle @ R.T

# Plot
plt.figure(figsize=(7, 7))
plt.plot(circle[:, 0], circle[:, 1], label="Original unit circle", linewidth=2)
plt.plot(rotated_circle[:, 0], rotated_circle[:, 1],
         label="Rotated circle", linestyle="--", linewidth=2)
plt.axhline(0, color="black", linewidth=0.7)
plt.axvline(0, color="black", linewidth=0.7)
plt.axis("equal")
plt.title(f"Rotation by {theta:.2f} radians")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Summary
- The inner product of functions is defined by an integral, and it measures similarity between functions.
- Orthogonal projection finds the closest approximation in a chosen subspace by minimizing squared error.
- Rotation preserves lengths and angles, and therefore preserves inner products.
- These ideas are central to approximation theory, Fourier analysis, data geometry, and machine learning.

Use the interval $[-\pi, \pi]$ and approximate the function

## Solution

First Solution

```python
# tha target function
f = x**2

# Basis functions
basis = [
    np.ones_like(x),
    np.cos(x),
    np.sin(x),
    np.cos(2*x),
    np.sin(2*x)
]

# Inner product formula: ci = <f, ei> / <ei, ei>
coefficients = [
    np.trapezoid(f * e, x) / np.trapezoid(e * e, x)
    for e in basis
]

for i, ci in enumerate(coefficients):
    print(f"c{i} = {ci:.6f}")
```

Second Solution

```python
import matplotlib.pyplot as plt

# Reconstruct approximation using projection coefficients
approximation = sum(
    ci * e for ci, e in zip(coefficients, basis)
)

# Plot original function and approximation
plt.figure(figsize=(10, 5))

plt.plot(x, f, label="Original f(x) = x²", linewidth=2)
plt.plot(x, approximation, "--", label="Approximation", linewidth=2)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Original Function vs Orthogonal Approximation")
plt.legend()
plt.grid(True)
plt.show()
```

Third solution

```python
# Compute squared approximation error
squared_error = (f - approximation) ** 2

# Integrate squared error over the interval
total_error = np.trapezoid(squared_error, x)

print(f"Squared Approximation Error: {total_error:.6f}")
```
